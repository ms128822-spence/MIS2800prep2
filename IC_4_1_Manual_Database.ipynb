```python
%%sql
/* 1. Create the table with STRICT types */
CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    category TEXT,
    price REAL
) STRICT;

/* 2. Insert at least 3 items */
INSERT INTO products (name, category, price) VALUES
('Superbowl Jersey', 'Clothing', 120.00),
('Football', 'Sports Equipment', 35.00),
('Team Hat', 'Accessories', 25.0);
```

```python
%%sql
/* Show all products that cost more than $50 */
SELECT *
FROM productsWHERE price > 50;
```

```python
%%sql
/* Update a price */
UPDATE products
SET price = 99.99
WHERE name = 'Superbowl Jersey';

/* Delete an item */
DELETE FROM products
WHERE product_id = 2;
```
